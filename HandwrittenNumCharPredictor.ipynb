```python
import tensorflow as tf
from tensorflow import keras
```

```python
import matplotlib.pyplot as plt
```

```python
(X_train, y_train) , (X_test, y_test) = tf.keras.datasets.mnist.load_data(0)
```

```python
print(X_train.shape)
print(X_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
for i in range(10):
  display(X_train[i])
```

```python
for i in range(10):
  display(y_train[i])
```

```python
X_train_flattened = X_train.reshape(len(X_train), 28*28)
display(X_train_flattened.shape)
X_test_flattened = X_test.reshape(len(X_test), 28*28)
display(X_test_flattened.shape)
```

```python
X_train_flattened = X_train_flattened/255
X_test_flattened = X_test_flattened/255
```

```python
model = tf.keras.Sequential([
  tf.keras.layers.Dense(128,input_shape=(784,), activation="relu"),
  tf.keras.layers.Dense(26, activation="softmax")
])
```

```python
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=['accuracy']
)
```

```python
history = model.fit(
    X_train_flattened, y_train,
    epochs=5,
    validation_data=(X_test_flattened, y_test)
)
```

```python
test_loss, test_acc = model.evaluate(X_test_flattened, y_test)
print(test_loss, test_acc)
```

```python
predictions = model.predict(X_test_flattened)
```

```python
import numpy as np
```

```python
# Look at the very first image in your test set (index 0)
first_prediction = predictions[0]

# np.argmax finds the index of the highest probability (e.g., 0 for A, 1 for B)
predicted_class = np.argmax(first_prediction)
actual_class = y_test[0]

print(f"The model guessed class: {predicted_class}")
print(f"The actual class was: {actual_class}")
```

```python
import matplotlib.pyplot as plt

# Plot training & validation accuracy values
plt.plot(history.history['accuracy'], label='Train Accuracy')
plt.plot(history.history['val_accuracy'], label='Validation Accuracy')
plt.title('Model Accuracy over Epochs')
plt.ylabel('Accuracy')
plt.xlabel('Epoch')
plt.legend(loc='lower right')
plt.show()
```

```python
from sklearn.metrics import confusion_matrix
import seaborn as sns
import numpy as np

# Get the raw predictions and convert them to actual class indices
predictions = model.predict(X_test_flattened)
predicted_classes = np.argmax(predictions, axis=1)

# Create the matrix
cm = confusion_matrix(y_test, predicted_classes)

# Plot it as a heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.xlabel('Predicted Character')
plt.ylabel('Actual Character')
plt.show()
```

```python
# Find indices where the prediction was wrong
incorrect_indices = np.where(predicted_classes != y_test)[0]

# Plot the first 5 mistakes
plt.figure(figsize=(15, 3))
for i, index in enumerate(incorrect_indices[:5]):
    plt.subplot(1, 5, i + 1)
    # Reshape the 784 array back into a 28x28 image for viewing
    plt.imshow(X_test[index].reshape(28, 28), cmap='gray')
    plt.title(f"Guessed: {predicted_classes[index]}\nActual: {y_test[index]}")
    plt.axis('off')
plt.show()
```
